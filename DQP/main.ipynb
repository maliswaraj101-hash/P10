```python
import pandas as pd  
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('train.csv')
```

```python
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
df.duplicated().sum()
```

```python
df['is_duplicate'].value_counts()
# slight imbalance
```

```python
new_df = df.sample(30000)
```

```python
new_df.isna().sum()
```

```python
new_df
```

```python
ques_df = new_df[['question1' , 'question2']]
```

```python
ques_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
questions = list(ques_df['question1']) + list(ques_df['question2'])
```

```python
questions
```

```python
cv = CountVectorizer(max_features=3000)
# 3000 most frequent used words will be used as a vocabulary
```

```python
q1_arr , q2_arr = np.vsplit(cv.fit_transform(questions).toarray() , 2)
# text representation of all our questions
```

```python
temp_df1 = pd.DataFrame(q1_arr , index=ques_df.index)
temp_df2 = pd.DataFrame(q2_arr , index=ques_df.index)

```

```python
temp_df = pd.concat([temp_df1 , temp_df2] , axis=1)
```

```python
temp_df.shape
```

```python
temp_df
```

```python
temp_df['is_duplicate'] = new_df['is_duplicate']
```

```python
temp_df
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(temp_df.iloc[: , 0:-1].values , temp_df.iloc[: , -1].values , test_size = 0.2 , random_state = 48)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
```

```python
rf.fit(x_train , y_train)
```

```python
y_pred = rf.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test , y_pred)
```
